# Exploring Relationships: Heatmaps with Python for Data Visualization.

```python
df.corr()
```

```python
plt.figure(figsize=(16,6))
sns.heatmap(df.corr(), annot=True)
plt.show()
```

# Numbers Affected and Unaffected by Heart Problems.

```python
df.columns
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    x='target',
    data=df,
    palette=['red', 'green']
)

plt.title('Count of Target')
plt.xlabel('Target')
plt.ylabel('Count')

plt.show()
```

# Gender Distribution In Dataset

```python
df.columns
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    x='sex',
    data=df,
    palette=['red', 'green']
)

plt.title('Count of Target')
plt.xlabel('Target')
plt.ylabel('Count')

plt.show()
```

# Distribution of Heart disease among Males and Females

```python
df.columns
```

```python
sns.countplot(data=df, x=df['sex'], hue=df['target'])
plt.xticks([0,1],['Female','Male'])
plt.legend(labels=['No-Disease','Disease'])
plt.show()
```

# Distribution of Ages in Our Dataset.

```python
df.columns
```

```python
sns.distplot(df['age'], bins=20)
plt.show()
```

# Chest Pain Types.

```python
df.columns
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x='cp', data=df, palette=['red', 'blue', 'green', 'orange'])

plt.title('Chest Pain Type Count')
plt.xlabel('Chest Pain Type')
plt.ylabel('Count')
plt.xticks([0,1,2,3],["Typical angina","Atypical angina","Non-anginal pain","Asymptomatic"])

plt.show()
```

# Chest Pain Distribution in Heart Disease vs Non-Heart Disease.

```python
sns.countplot(data=df, x='cp', hue='target')
plt.legend(labels=['Non_Disease','Disease'])
plt.show()
```
