```python
import torch
```

```python
t_u = torch.tensor([35.7, 55.9, 58.2, 81.9])
t_c = torch.tensor([0.5, 14.0, 15.0, 28.0])
```

```python
w = torch.tensor(0.5, requires_grad=True)
b = torch.tensor(-17.0, requires_grad=True)
```

```python
def model(t_u, w, b):
    return w * t_u + b
```

```python
def loss_fn(t_p, t_c):
    return ((t_p - t_c) ** 2).mean()
```

```python
t_p = model(t_u, w, b)
```

```python
loss = loss_fn(t_p, t_c)
```

```python
print("Loss:", loss)
```

```python
print("t_p:",t_p)
```
