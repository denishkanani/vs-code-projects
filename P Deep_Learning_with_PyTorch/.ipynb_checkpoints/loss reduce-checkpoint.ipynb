```python
import torch
import matplotlib.pyplot as plt

# Data
X = torch.tensor([1000.,1500.,2000.,2500.,3000.])
Y = torch.tensor([20.,30.,40.,50.,60.])

# Normalize
X = X / 1000

# Parameters
w = torch.tensor(0.5, requires_grad=True)
b = torch.tensor(-17., requires_grad=True)

# Learning rate
lr = 0.01

# Training
for epoch in range(2925):

    prediction = w * X + b

    loss = ((prediction - Y)**2).mean()

    loss.backward()

    with torch.no_grad():
        w -= lr * w.grad
        b -= lr * b.grad

        w.grad.zero_()
        b.grad.zero_()

print("Weight:", w.item())
print("Bias:", b.item())
print("Loss:", loss.item())

# Plot
prediction = w * X + b

plt.scatter(X, Y, label="Actual Data")
plt.plot(X, prediction, color="red", label="Prediction")
plt.legend()
plt.show()
```

```python
import torch

# Data
X = torch.tensor([1.,2.,3.,4.])
Y = torch.tensor([3.,5.,7.,9.])

# Parameters (initial guesses)
w = torch.tensor(0.5, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)

# Learning rate
lr = 0.1

# Model
def model(x):
    return w * x + b

# Loss function
def loss_fn(pred, actual):
    return ((pred - actual) ** 2).mean()

# Training loop
for epoch in range(20):

    # Step 1: Prediction
    pred = model(X)

    # Step 2: Calculate loss
    loss = loss_fn(pred, Y)

    # Step 3: Calculate gradients
    loss.backward()

    # Step 4: Update parameters
    with torch.no_grad():
        w -= lr * w.grad
        b -= lr * b.grad

        # Step 5: Clear gradients
        w.grad.zero_()
        b.grad.zero_()

    print(f"Epoch {epoch+1}")
    print(f"Weight = {w.item():.4f}")
    print(f"Bias = {b.item():.4f}")
    print(f"Loss = {loss.item():.4f}")
    print("----------------------")
```
