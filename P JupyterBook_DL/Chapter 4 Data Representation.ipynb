```python
import torch

data = torch.tensor([25,26,27,28,29,30])

window = 3

for i in range(len(data)-window):
    x = data[i:i+window]
    y = data[i+window]
    print(x, "->", y)
```

## Full Code

```python
import torch
import torch.nn as nn
import numpy as np

# ============================================
# 1. IMAGE
# ============================================

image = torch.randn(3, 64, 64)

print("Image:", image.shape)

# Add batch dimension
image_batch = image.unsqueeze(0)

print("Image batch:", image_batch.shape)


# ============================================
# 2. IMAGE NORMALIZATION
# ============================================

image = image.float()

image = (image - image.mean()) / image.std()

print("Normalized image mean:", image.mean())
print("Normalized image std:", image.std())


# ============================================
# 3. 3D VOLUME
# ============================================

volume = torch.randn(
    64, 128, 128
)

print("Volume:", volume.shape)


# ============================================
# 4. TABULAR DATA
# ============================================

data = torch.tensor([
    [20., 170., 65.],
    [25., 175., 72.],
    [30., 180., 80.]
])

print("Tabular:", data.shape)


# ============================================
# 5. CATEGORICAL DATA
# ============================================

labels = torch.tensor([
    0, 1, 2, 1
])

one_hot = torch.nn.functional.one_hot(
    labels,
    num_classes=3
)

print("One hot:")
print(one_hot)


# ============================================
# 6. TIME SERIES
# ============================================

values = torch.tensor([
    10., 20., 30., 40., 50., 60.
])

window = 3

X = []
y = []

for i in range(len(values) - window):
    X.append(values[i:i + window])
    y.append(values[i + window])

X = torch.stack(X)
y = torch.stack(y)

print("Time-series X:")
print(X)

print("Time-series y:")
print(y)


# ============================================
# 7. TEXT
# ============================================

vocab = {
    "hello": 0,
    "world": 1,
    "deep": 2,
    "learning": 3
}

sentence = "hello deep learning"

tokens = [
    vocab[word]
    for word in sentence.split()
]

tokens = torch.tensor(tokens)

print("Token IDs:")
print(tokens)


# ============================================
# 8. TEXT EMBEDDING
# ============================================

embedding = nn.Embedding(
    num_embeddings=4,
    embedding_dim=8
)

embedded = embedding(tokens)

print("Embedding shape:")
print(embedded.shape)
```
